## Import a catalog
> The `Visualens.import_catalog()` function allows the user to import astronomical catalogs. The input can be the path to a fits or raw text (csv) catalog, or a `astropoy.table.Table` object directly.

### Selection panel and red sequence

- Creates a panel with a scatter plot of objects according to two specified columns:
> `vl.catalog.make_selection_panel()`

> The selection is saved in `vl.catalog.selection_mask`. This is particularly useful when making a red sequence selection of cluster galaxies for example. You can save the selection directly with `vl.catalog.export_to_potfile(path)`.

- Creates a panel with a scatter plot of objects according to two specified columns:
> Close the panel with `vl.catalog.remove_selection_panel()`

```python
# Create a new column for a specific quantity we want to look at. Here, color.
vl.catalog.cat['f115w_mag-f200w_mag'] = vl.catalog.cat['f115w_mag'] - vl.catalog.cat['f200w_mag']

# Create the selection panel
vl.catalog.make_selection_panel(xy_axes=['f200w_mag', 'f115w_mag-f200w_mag'])
```

## Import a lens model

`visualens` uses `Lenstool`'s python wrapper as its lensing engine and takes as inputs lens model files in the `Lenstool` format.

- To import your lens model (`model_path` can be the path to the directory containing your lens model's files, or the path to any of these files):
> `vl.import_lens_model(model_path)`

**Careful:** This will move the current working directory to your lens model's directory, so make sure to use absolute paths.

All lensing functions are now accessible in the `vl.lens_model`. **`visualens` automatically calculates the sources and predicted image positions from the multiple images, as well as lensing quantities like magnification, shear, etc.**, so this step can take a little bit to run if you have many multiple images (a few seconds, typically less than a minute).


> **Note:** `Lenstool`'s python wrapper will overwrite some of your output files (`image.all`, `ce.dat`, etc.) as it recomputes the corresponding data.

```python
model_path = DATA_dir + "/lens_model/"
vl.import_lenstool(model_path)
```

## Set the source plane's redshift

- Many functions of `vl.lens_model` run calculations in the source plane. To set the source plane's redshift:
> `vl.lens_model.set_lt_z(redshift)`

> `visualens` will automatically compute the critical and caustic curves at the given redshift, which can take a little bit of time. One it has computed them, it saves them in `lt_curves.pkl` and `lt_caustics.pkl` to avoid recalculating them later.

```python
vl.lens_model.set_lt_z(6.2)
```

## Transfer data from the imported catalog

You can transfer data from the imported catalog to the multiple images and other lensing catalogs.

- To transfer a column from the most recently imported catalog (`vl.catalog`):
> `Image.lt.mult.transfer_col(column_name)`, accessible through `vl.lens_model.mult.transfer_col(column_name)`

- To transfer from an earlier imported catalog, pass its index in `vl.catalogs`:
> `vl.lens_model.mult.transfer_col(column_name, index=0)`

```python
vl.lens_model.mult.transfer_col('z_phot')
```

```python
print(vl.lens_model.mult.cat.colnames)
```
